```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns

# Set up visualization style
plt.style.use('default')
sns.set_theme()
```

### 1. Read Spotify data
##### Read the Spotify artist data

```python
spotify_df = pd.read_csv('spotify_artist_data.csv')

# Display basic information about the dataframe
print("Spotify data shape:", spotify_df.shape)
print("\nColumns:", spotify_df.columns.tolist())
print("\nFirst few rows:")
spotify_df.head()
```

### 2. Read Last.fm data
##### Read the Last.fm artist data


```python
lastfm_df = pd.read_csv('lastfm_artist_data.csv')

# Display basic information about the dataframe
print("Last.fm data shape:", lastfm_df.shape)
print("\nColumns:", lastfm_df.columns.tolist())
print("\nFirst few rows:")
lastfm_df.head()
```

### 3. Process JSON columns

```python
# Convert JSON strings to Python objects for Spotify data
spotify_df['top_tracks_parsed'] = spotify_df['top_tracks'].apply(lambda x: json.loads(x) if pd.notna(x) and x != '' else [])
spotify_df['genres_list'] = spotify_df['genres'].apply(lambda x: x.split('|') if pd.notna(x) and x != '' else [])

# Convert JSON strings for Last.fm data
lastfm_df['top_tracks_parsed'] = lastfm_df['top_tracks'].apply(lambda x: json.loads(x) if pd.notna(x) and x != '' else [])
lastfm_df['top_albums_parsed'] = lastfm_df['top_albums'].apply(lambda x: json.loads(x) if pd.notna(x) and x != '' else [])
lastfm_df['similar_artists_parsed'] = lastfm_df['similar_artists'].apply(lambda x: json.loads(x) if pd.notna(x) and x != '' else [])
lastfm_df['tags_list'] = lastfm_df['tags'].apply(lambda x: x.split('|') if pd.notna(x) and x != '' else [])
```

### 4. Data exploration examples

```python

# Example 1: Most popular artists on Spotify
top_spotify_artists = spotify_df.nlargest(10, 'popularity')
plt.figure(figsize=(12, 6))
plt.barh(top_spotify_artists['artist_name'], top_spotify_artists['popularity'])
plt.xlabel('Popularity Score')
plt.title('Top 10 Most Popular Artists on Spotify')
plt.tight_layout()
plt.show()
```

```python
# Example 2: Artists with most followers
top_follower_artists = spotify_df.nlargest(10, 'followers')
plt.figure(figsize=(12, 6))
plt.barh(top_follower_artists['artist_name'], top_follower_artists['followers'])
plt.xlabel('Number of Followers')
plt.title('Top 10 Artists by Followers')
plt.tight_layout()
plt.show()
```

```python
# Example 3: Genre distribution
all_genres = []
for genres in spotify_df['genres_list']:
    all_genres.extend(genres)

genre_counts = pd.Series(all_genres).value_counts()
plt.figure(figsize=(12, 8))
genre_counts.head(20).plot(kind='bar')
plt.title('Top 20 Genres Distribution')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Example 4: Correlation between Last.fm and Spotify popularity (if we merge the data)
merged_df = pd.merge(spotify_df, lastfm_df, on='artist_id', how='inner', suffixes=('_spotify', '_lastfm'))

plt.figure(figsize=(10, 8))
plt.scatter(merged_df['popularity'], merged_df['playcount'], alpha=0.6)
plt.xlabel('Spotify Popularity Score')
plt.ylabel('Last.fm Play Count')
plt.title('Spotify Popularity vs Last.fm Play Count')
plt.xscale('log')
plt.yscale('log')
plt.tight_layout()
plt.show()
```

### 5. Helper functions for data analysis

```python
def get_top_tracks_by_artist(artist_name, df):
    """Get top tracks for a specific artist"""
    artist_data = df[df['artist_name'] == artist_name].iloc[0]
    tracks = artist_data['top_tracks_parsed']
    if tracks:
        return pd.DataFrame(tracks)
    return None

def get_genre_summary(df):
    """Get summary statistics by genre"""
    # Create a new dataframe with one row per artist-genre combination
    genre_rows = []
    for _, row in df.iterrows():
        for genre in row['genres_list']:
            genre_rows.append({
                'genre': genre,
                'artist_id': row['artist_id'],
                'artist_name': row['artist_name'],
                'popularity': row['popularity'],
                'followers': row['followers']
            })
    
    genre_df = pd.DataFrame(genre_rows)
    
    # Calculate summary statistics
    summary = genre_df.groupby('genre').agg({
        'artist_id': 'count',
        'popularity': 'mean',
        'followers': 'mean'
    }).rename(columns={'artist_id': 'artist_count'})
    
    return summary.sort_values('artist_count', ascending=False)

# Example usage of helper functions
genre_summary = get_genre_summary(spotify_df)
print("\nGenre Summary (Top 10):")
print(genre_summary.head(10))

# Get top tracks for a specific artist
artist_name = spotify_df.iloc[0]['artist_name']
top_tracks = get_top_tracks_by_artist(artist_name, spotify_df)
if top_tracks is not None:
    print(f"\nTop tracks for {artist_name}:")
    print(top_tracks)
```
